# Data Wrangling

In the previous Exploratory Data Analysis phase, we began showing some of the typical data challenges we confront every day.  Hopefully you're starting to understand there's a significant amount of time devoted to acquiring the data, analyzing, summarizing, asking questions, and digging in until it makes sense.  Real-world data can be complicated and there's no one size fits all methodology.  Nearly every dataset will be unique and have peculiarities specific to the domain and data collections method.

_Data Integrity_ addresses any issues contained in the data, and ensures your data is accurate, complete, and ready for analysis.  Remember the old adage, "garbage in, garbage out".  If your data is riddled with holes and untrustworthy data, then you can count on your analysis being suspect too.  So again, be sure to give this EDA step all the attention it requires.
